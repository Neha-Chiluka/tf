```python
import tensorflow as tf
```

```python
df = tf.data.experimental.make_csv_dataset(r'/content/sample_data/dataset/drugLibTrain_raw.tsv',
                                           batch_size=1, field_delim='\t')
```

```python
def prep_ds(ds, shuffle_buffer_size=1024, \
                batch_size=32):
        # Shuffle the dataset
        ds = ds.shuffle(buffer_size=shuffle_buffer_size)
        # Repeat the dataset
        ds = ds.repeat()
        # Batch the dataset
        ds = ds.batch(batch_size)
        return ds
```

```python
ds = prep_ds(df, batch_size=5)
```

```python

for x in ds.take(1):\
        print(x)
```

```python
import tensorflow_hub as hub
embedding = "https://tfhub.dev/google/tf2-preview"\
                "/gnews-swivel-20dim/1"
hub_layer = hub.KerasLayer(embedding, input_shape=[], \
                               dtype=tf.string, \
                               trainable=True)
```
