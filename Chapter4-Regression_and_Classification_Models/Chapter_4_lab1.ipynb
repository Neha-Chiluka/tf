```python
import tensorflow as tf

model = tf.keras.Sequential()
model.add(tf.keras.layers.InputLayer(input_shape=(10,), name='Input_layer_1'))
# This is your initial InputLayer, defining the model's input shape

# Instead of adding another InputLayer, add subsequent layers directly
# For example:
model.add(tf.keras.layers.Dense(32, activation='relu'))
model.add(tf.keras.layers.Dense(10, activation='softmax'))
```
